```python
def tenure_group(t):
    if t<=12:
        return 'Short'
    elif t<=48:
        return 'Medium'
    else:
        return 'Long'
df_encod['tenure_group']=df_encod['tenure'].apply(tenure_group)
```

```python
df_encod.head(2)
```

```python
df_encod=pd.get_dummies(df_encod,columns=['tenure_group'],drop_first=True,dtype=int)
df_encod.head(2)
```

```python
df_encod['AvgMonthSpend']=(df_encod['MonthlyCharges']*df_encod['tenure'])/(df_encod['tenure']+1)
```

```python
df_encod.head(2)
```

```python
from sklearn.model_selection import train_test_split

X=df_encod.drop('Churn',axis=1)

Y=df_encod['Churn']

X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)

print("Training data shape:",X_train.shape)
print("Testing data shape:",X_test.shape)




```

```python
#logistic regression
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix

log_model=LogisticRegression(max_iter=1000)
log_model.fit(X_train,Y_train)

Y_pred_log=log_model.predict(X_test)

print("Logistic Regression Accuracy:",accuracy_score(Y_test,Y_pred_log))
print(confusion_matrix(Y_test,Y_pred_log))
print(classification_report(Y_test,Y_pred_log))
```

```python
#decision tree
from sklearn.tree import DecisionTreeClassifier

tree_model=DecisionTreeClassifier(random_state=42)
tree_model.fit(X_train,Y_train)

Y_pred_tree=tree_model.predict(X_test)


print("Decision Tree Accuracy:",accuracy_score(Y_test,Y_pred_tree))
print(confusion_matrix(Y_test,Y_pred_tree))
print(classification_report(Y_test,Y_pred_tree))

    
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn_model=KNeighborsClassifier(n_neighbors=5)
knn_model.fit(X_train,Y_train)

Y_pred_knn=knn_model.predict(X_test)


print("KNN Accuracy:",accuracy_score(Y_test,Y_pred_knn))
print(confusion_matrix(Y_test,Y_pred_knn))
print(classification_report(Y_test,Y_pred_knn))
```

# Project Summary: Customer Churn Prediction

## Objective
Predict telecom customer churn and identify key factors influencing customer retention.

## Work Done
- Cleaned and prepared raw telecom dataset
- Handled missing values and outliers
- Performed feature encoding and scaling
- Balanced dataset using SMOTE
- Built ML models:
  - Logistic Regression
  - Decision Tree
  - KNN
- Evaluated models using Accuracy, Confusion Matrix, Precision, Recall
- Applied K-Means clustering for customer segmentation

## Results
- Best model accuracy: 81.62%
- Key churn drivers:
  - Month-to-month contract
  - High monthly charges
  - Low tenure

## Business Insights
- Offer discounts for high-charge customers
- Encourage long-term contracts
- Improve retention for new customers
- Use segmentation to target high-risk groups

## Outcome
Developed an end-to-end machine learning pipeline for churn prediction and customer segmentation.
